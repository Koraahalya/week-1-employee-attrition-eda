## 2. Data Cleaning

### 2.1 Missing Value Analysis

Missing values are checked before performing further analysis.
Missing values can affect statistical calculations and machine learning
models, so they must be identified and handled appropriately.

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
missing_counts = df.isnull().sum()

missing_counts = missing_counts[missing_counts > 0]

plt.figure(figsize=(10, 5))

if len(missing_counts) == 0:
    plt.text(
        0.5,
        0.5,
        "No Missing Values Found",
        ha="center",
        va="center",
        fontsize=16
    )
    plt.axis("off")
else:
    missing_counts.sort_values(ascending=False).plot(kind="bar")
    plt.title("Missing Values by Column")
    plt.xlabel("Columns")
    plt.ylabel("Number of Missing Values")
    plt.xticks(rotation=90)

plt.tight_layout()
plt.savefig("missing_values.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
print("Number of unique values in each column:")
display(df.nunique().sort_values())
```

```python
constant_columns = []

for column in df.columns:
    if df[column].nunique() == 1:
        constant_columns.append(column)

print("Columns containing only one unique value:")
print(constant_columns)
```

```python
df_clean = df.copy()

print("Original dataset shape:", df.shape)
print("Cleaning copy created successfully.")
```

```python
constant_columns = [
    "EmployeeCount",
    "Over18",
    "StandardHours"
]

df_clean = df_clean.drop(columns=constant_columns)

print("Removed constant columns:")
print(constant_columns)

print("Dataset shape after removing constant columns:", df_clean.shape)
```

```python
df_clean = df_clean.drop(columns=["EmployeeNumber"])

print("EmployeeNumber removed because it is an identifier.")
print("Dataset shape after removing EmployeeNumber:", df_clean.shape)
```

```python
print("Cleaned dataset shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum().sum())

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())
```

```python
print("Data types after cleaning:")
display(df_clean.dtypes)
```

```python
numerical_columns = df_clean.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df_clean.select_dtypes(include="object").columns.tolist()

print("Number of numerical columns:", len(numerical_columns))
print("\nNumerical columns:")
print(numerical_columns)

print("\nNumber of categorical columns:", len(categorical_columns))
print("\nCategorical columns:")
print(categorical_columns)
```

```python
print("Attrition value counts:")
display(df_clean["Attrition"].value_counts())

print("\nAttrition percentages:")
attrition_percentage = df_clean["Attrition"].value_counts(normalize=True) * 100
display(attrition_percentage.round(2))
```

```python
plt.figure(figsize=(7, 5))

ax = sns.countplot(data=df_clean, x="Attrition")

plt.title("Employee Attrition Distribution")
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.savefig("01_attrition_distribution.png", dpi=300, bbox_inches="tight")
plt.show()
```

### Insight

The dataset contains substantially more employees who did not leave the
organization than employees who left. Therefore, the Attrition variable
is imbalanced.

This imbalance is important for future predictive modelling because a model
could achieve high overall accuracy by mainly predicting the majority class.
Therefore, metrics such as precision, recall, F1-score, and ROC-AUC should
also be considered during future modelling.

```python
plt.figure(figsize=(10, 6))

ax = sns.countplot(
    data=df_clean,
    x="Department",
    hue="Attrition"
)

plt.title("Employee Attrition by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=15)

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.savefig("02_attrition_by_department.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
department_attrition = pd.crosstab(
    df_clean["Department"],
    df_clean["Attrition"]
)

print("Department-wise Attrition Counts:")
display(department_attrition)
```

```python
department_attrition_rate = pd.crosstab(
    df_clean["Department"],
    df_clean["Attrition"],
    normalize="index"
) * 100

print("Department-wise Attrition Percentage:")
display(department_attrition_rate.round(2))
```

### Insight

The department visualization shows the number of employees who stayed and
left in each department.

However, raw employee counts should not be interpreted as attrition risk
because departments have different numbers of employees. Therefore,
department-wise attrition percentages were calculated to provide a fairer
comparison.

The department with the highest number of employees leaving is not
necessarily the department with the highest attrition rate. The percentage
table should be used to identify departments with relatively higher
attrition risk.

```python
plt.figure(figsize=(8, 5))

ax = sns.countplot(
    data=df_clean,
    x="OverTime",
    hue="Attrition"
)

plt.title("Employee Attrition by Overtime Status")
plt.xlabel("Overtime")
plt.ylabel("Number of Employees")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()
plt.savefig("03_overtime_vs_attrition.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
overtime_attrition = pd.crosstab(
    df_clean["OverTime"],
    df_clean["Attrition"]
)

print("Overtime vs Attrition:")
display(overtime_attrition)
```

```python
overtime_attrition_rate = pd.crosstab(
    df_clean["OverTime"],
    df_clean["Attrition"],
    normalize="index"
) * 100

print("Overtime-wise Attrition Percentage:")
display(overtime_attrition_rate.round(2))
```

### Insight

The analysis indicates a noticeable difference in attrition between
employees who work overtime and those who do not.

Employees working overtime show a higher proportion of attrition compared
with employees who do not work overtime. This suggests that overtime status
may be associated with employee attrition.

However, this analysis only identifies an association. It does not prove
that overtime directly causes employees to leave because other factors may
also influence attrition.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df_clean,
    x="Attrition",
    y="MonthlyIncome"
)

plt.title("Monthly Income Distribution by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")

plt.tight_layout()
plt.savefig("04_monthly_income_vs_attrition.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
income_summary = df_clean.groupby("Attrition")["MonthlyIncome"].agg(
    ["count", "mean", "median", "min", "max"]
)

print("Monthly Income Summary by Attrition:")
display(income_summary.round(2))
```

### Insight

The boxplot shows differences in the distribution of monthly income between
employees who left and employees who remained.

The mean and median values provide a numerical comparison of the two groups.
The boxplot also shows the spread of income and the presence of high-income
outliers.

This suggests that income level may be related to employee attrition.
However, income should not be considered independently because factors such
as job level, experience, job role, and years at the company can also affect
monthly income and attrition.

```python
numeric_df = df_clean.select_dtypes(include=np.number)

correlation_matrix = numeric_df.corr()

plt.figure(figsize=(16, 12))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Numerical Variables")

plt.tight_layout()
plt.savefig("05_correlation_heatmap.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
corr_pairs = correlation_matrix.unstack()

corr_pairs = corr_pairs[
    corr_pairs.index.get_level_values(0) != corr_pairs.index.get_level_values(1)
]

corr_pairs = corr_pairs.sort_values(key=abs, ascending=False)

print("Strongest correlations:")
display(corr_pairs.head(20))
```

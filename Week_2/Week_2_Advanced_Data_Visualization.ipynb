### Insight

The department visualization shows the number of employees who stayed and
left in each department.

However, raw employee counts should not be interpreted as attrition risk
because departments have different numbers of employees. Therefore,
department-wise attrition percentages were calculated to provide a fairer
comparison.

The department with the highest number of employees leaving is not
necessarily the department with the highest attrition rate. The percentage
table should be used to identify departments with relatively higher
attrition risk.

```python
# ============================================================
# WEEK 2 - ADVANCED DATA VISUALIZATION AND STORYTELLING
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load the cleaned Week 1 dataset
df_clean = pd.read_csv("cleaned_employee_attrition.csv")

print("Dataset loaded successfully.")
print("Shape:", df_clean.shape)
print("\nColumns:")
print(df_clean.columns.tolist())

print("\nFirst 5 rows:")
display(df_clean.head())
```

```python
# Check that important columns required for Week 2 are available

required_columns = [
    "Attrition",
    "Age",
    "Department",
    "JobRole",
    "JobLevel",
    "MonthlyIncome",
    "OverTime",
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "RelationshipSatisfaction",
    "WorkLifeBalance",
    "YearsAtCompany"
]

missing_required = [
    col for col in required_columns
    if col not in df_clean.columns
]

if missing_required:
    print("Missing required columns:")
    print(missing_required)
else:
    print("All required Week 2 columns are available.")

print("\nAttrition values:")
print(df_clean["Attrition"].value_counts())
```

```python
# ============================================================
# ATTRITION RATE BY AGE GROUP
# ============================================================

# Create meaningful age groups
df_clean["AgeGroup"] = pd.cut(
    df_clean["Age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56+"],
    include_lowest=True
)

# Calculate attrition rate
age_attrition = (
    df_clean.groupby("AgeGroup", observed=False)["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="AttritionRate")
)

# Plot
plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=age_attrition,
    x="AgeGroup",
    y="AttritionRate"
)

plt.title(
    "Employee Attrition Rate by Age Group",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")

# Add values to bars
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=4
    )

plt.tight_layout()

plt.savefig(
    "08_age_group_attrition_rate.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("""
Narrative:
This visualization compares employee attrition rates across different
age groups. Using percentages instead of employee counts allows fair
comparison between groups of different sizes. It helps identify whether
employee turnover is concentrated among younger or older employees.
""")
```

```python
# ============================================================
# ATTRITION RATE BY JOB ROLE
# ============================================================

role_attrition = (
    df_clean.groupby("JobRole")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="AttritionRate")
    .sort_values("AttritionRate", ascending=True)
)

plt.figure(figsize=(11, 7))

ax = sns.barplot(
    data=role_attrition,
    x="AttritionRate",
    y="JobRole"
)

plt.title(
    "Employee Attrition Rate by Job Role",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel("Attrition Rate (%)")
plt.ylabel("Job Role")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=4
    )

plt.tight_layout()

plt.savefig(
    "09_job_role_attrition_rate.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("""
Narrative:
This visualization identifies job roles with comparatively high and
low attrition rates. A role-level view is useful because overall
attrition can hide important differences between types of employees.
The result can help HR identify roles that may need retention analysis.
""")
```

```python
# ============================================================
# MONTHLY INCOME BY JOB LEVEL AND ATTRITION
# ============================================================

plt.figure(figsize=(11, 7))

sns.violinplot(
    data=df_clean,
    x="JobLevel",
    y="MonthlyIncome",
    hue="Attrition",
    split=True,
    inner="quartile"
)

plt.title(
    "Monthly Income Distribution by Job Level and Attrition",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Job Level")
plt.ylabel("Monthly Income")

plt.legend(
    title="Attrition",
    loc="upper right"
)

plt.tight_layout()

plt.savefig(
    "10_monthly_income_joblevel_attrition.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("""
Narrative:
The violin plot compares the distribution of monthly income for
employees who stayed and employees who left, across job levels.
It shows not only the median and quartiles, but also the shape and
density of income distributions. This provides a deeper view of the
relationship between compensation, job level and attrition.
""")
```

```python
# ============================================================
# SATISFACTION FACTORS AND ATTRITION RATE
# ============================================================

satisfaction_columns = [
    "JobSatisfaction",
    "EnvironmentSatisfaction",
    "RelationshipSatisfaction",
    "WorkLifeBalance"
]

satisfaction_results = []

for column in satisfaction_columns:

    temp = (
        df_clean.groupby(column)["Attrition"]
        .apply(lambda x: (x == "Yes").mean() * 100)
        .reset_index(name="AttritionRate")
    )

    temp["Factor"] = column

    satisfaction_results.append(temp)

satisfaction_df = pd.concat(
    satisfaction_results,
    ignore_index=True
)

# Convert to a wide table
satisfaction_heatmap = satisfaction_df.pivot_table(
    index="Factor",
    columns=satisfaction_df.columns[0],
    values="AttritionRate",
    aggfunc="mean"
)

plt.figure(figsize=(10, 6))

sns.heatmap(
    satisfaction_heatmap,
    annot=True,
    fmt=".1f",
    cmap="YlOrRd",
    linewidths=0.5,
    cbar_kws={"label": "Attrition Rate (%)"}
)

plt.title(
    "Attrition Rate Across Employee Satisfaction Factors",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Satisfaction / Balance Level")
plt.ylabel("Factor")

plt.tight_layout()

plt.savefig(
    "11_satisfaction_attrition_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("""
Narrative:
This heatmap examines attrition rates at different levels of job
satisfaction, environmental satisfaction, relationship satisfaction,
and work-life balance. It provides a compact comparison of several
employee experience factors at the same time and helps identify
patterns that deserve further investigation.
""")
```

```python
# ============================================================
# OVERTIME AND JOB LEVEL VS ATTRITION
# ============================================================

overtime_joblevel = (
    df_clean.groupby(
        ["OverTime", "JobLevel"]
    )["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
    .reset_index(name="AttritionRate")
)

overtime_heatmap = overtime_joblevel.pivot(
    index="OverTime",
    columns="JobLevel",
    values="AttritionRate"
)

plt.figure(figsize=(10, 5))

sns.heatmap(
    overtime_heatmap,
    annot=True,
    fmt=".1f",
    cmap="Reds",
    linewidths=0.5,
    cbar_kws={"label": "Attrition Rate (%)"}
)

plt.title(
    "Attrition Rate by Overtime Status and Job Level",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Job Level")
plt.ylabel("Overtime Status")

plt.tight_layout()

plt.savefig(
    "12_overtime_joblevel_attrition_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("""
Narrative:
This heatmap investigates the combined relationship between overtime,
job level and employee attrition. Looking at these variables together
is more informative than examining overtime alone because the effect
of overtime may differ across employee seniority levels.
""")
```

```python
# ============================================================
# WEEK 2 - ANALYTICAL INSIGHTS
# ============================================================

print("=" * 70)
print("WEEK 2 EXPLORATORY INSIGHTS")
print("=" * 70)

# Overall attrition rate
overall_attrition = (
    df_clean["Attrition"].eq("Yes").mean() * 100
)

print(f"\n1. Overall Attrition Rate: {overall_attrition:.2f}%")

# Highest attrition age group
highest_age = age_attrition.loc[
    age_attrition["AttritionRate"].idxmax()
]

print(
    f"2. Highest Age-Group Attrition: "
    f"{highest_age['AgeGroup']} "
    f"({highest_age['AttritionRate']:.2f}%)"
)

# Highest attrition job role
highest_role = role_attrition.iloc[-1]

print(
    f"3. Highest Job-Role Attrition: "
    f"{highest_role['JobRole']} "
    f"({highest_role['AttritionRate']:.2f}%)"
)

# Overtime attrition
overtime_rates = (
    df_clean.groupby("OverTime")["Attrition"]
    .apply(lambda x: (x == "Yes").mean() * 100)
)

highest_overtime = overtime_rates.idxmax()

print(
    f"4. Higher Overtime Attrition: "
    f"{highest_overtime} "
    f"({overtime_rates[highest_overtime]:.2f}%)"
)

# Average monthly income by attrition
income_summary = (
    df_clean.groupby("Attrition")["MonthlyIncome"]
    .mean()
)

print(
    f"5. Average Monthly Income of Employees Who Left: "
    f"{income_summary.get('Yes', np.nan):.2f}"
)

print(
    f"6. Average Monthly Income of Employees Who Stayed: "
    f"{income_summary.get('No', np.nan):.2f}"
)

print("\nKey observation:")
print(
    "The analysis suggests that employee attrition is not driven by "
    "one single variable. Age, job role, compensation, satisfaction, "
    "job level and overtime provide different dimensions for understanding "
    "employee turnover."
)
```
